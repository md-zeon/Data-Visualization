```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
```

```python
print(tips.head())
```

```python
sns.set_theme(style="ticks")
sns.relplot(data=tips, x="total_bill", y="tip", hue="smoker", size="size", style="sex")
```

```python
x = np.arange(0, 11);
y = x ** 2;
sns.set_theme(style="whitegrid")
sns.relplot(x=x, y=y, kind="line")
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
```

```python
sns.lineplot(data=tips, x="total_bill", y="tip", hue="time")
```

```python
flights = sns.load_dataset("flights")
flights
```

```python
sns.lineplot(data=flights, x="year", y="passengers")
```

```python
sns.barplot(data=tips, x="day", y="total_bill", hue="sex")
```

```python
sns.boxplot(data=tips, x="day", y="tip")
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(data=penguins, x="body_mass_g", bins=20)
```

```python
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")
```

```python
sns.heatmap(data=flights_pivot, cmap="coolwarm", annot=True, fmt="d")
plt.title("Number of Passengers per Month and Year")
plt.tight_layout()
```

```python
tips = sns.load_dataset("tips")
fig, ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
    marker="o",
    errorbar=None
)
ax.set_title("Total Bill by Day and Sex")
ax.set_xlabel("Day of the Week")
ax.set_ylabel("Total Bill ($)")

fig.tight_layout()
```

```python
oscar_movies = [
    "The Dark Knight",
    "The Heart Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in M $

plt.plot(years, oscar_revenue, marker='o')
plt.title("Oscar Winning Movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (M $)")
```

```python
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [170, 2788, 829, 185, 275] # in M $

plt.plot(years, oscar_revenue, label="Oscar Winning Movies", marker='o')
plt.plot(years, non_oscar_revenue, label="Non-Oscar Winning Movies", marker='o')
plt.title("Oscar vs Non-Oscar Winning Movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (M $)")
plt.legend()
```

```python
people = ["Person A", "Person B", "Person C", "Person D", "Person E", "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
colors = ["green" if x < 135 else "red" for x in blood_pressure]
sizes = np.array(blood_pressure)
# plt.scatter(age, blood_pressure, color=colors, s=sizes/2.5, marker=".")
plt.scatter(age, blood_pressure, s=sizes/2, cmap='OrRd', c=blood_pressure)
plt.grid(linestyle='--', alpha=0.7)

plt.colorbar(label="Blood Pressure")

for i in range(len(age)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i]), xytext=(age[i],blood_pressure[i]+2))
plt.xlim(min(age), max(age)+10)
plt.ylim(min(blood_pressure), max(blood_pressure)+5)
plt.title("Age vs Blood Pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# Stack Plots

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [20, 35, 30, 35, 27, 25, 32]
organic = [15, 25, 20, 30, 22, 20, 28]
social = [10, 15, 12, 18, 14, 10, 20]

# Single day pie chart
# plt.pie([direct[0], organic[0], social[0]], labels=['Direct', 'Organic', 'Social'], autopct='%1.1f%%', startangle=90)

# Multi-day stack plot
plt.stackplot(days, direct, organic, social, labels=['Direct', 'Organic', 'Social'], colors=['#ff9999','#66b3ff','#99ff99'])
plt.title("Marketing Traffic Sources Over a Week")
plt.xlabel("Days of the Week")
plt.ylabel("Number of Customers")
plt.legend()
plt.show()
```

```python
# Sub Plots
x = np.arange(1, 11, 1)
y1 = np.sqrt(x)
y2 = np.double(x)
y3 = np.square(x)
y4 = x ** 3

plt.subplot(2, 2, 1)
plt.plot(x, y1, label='y = sqrt(x)', color='blue')
plt.title("Plot of Square Root")

plt.subplot(2, 2, 2)
plt.plot(x, y2, label='y = 2x', color='orange')
plt.title("Plot of Double")

plt.subplot(2, 2, 3)
plt.plot(x, y3, label='y = x^2', color='green')
plt.title("Plot of Square")

plt.subplot(2, 2, 4)
plt.plot(x, y4, label='y = x^3', color='red')
plt.title("Plot of Cube")

plt.tight_layout()
plt.show()
```

```python
fig, axis = plt.subplots(2, 2)

axis[0, 0].plot(x, y1, label='y = sqrt(x)', color='blue')
axis[0, 0].set_title("Plot of Square Root")

axis[0, 1].plot(x, y2, label='y = 2x', color='orange')
axis[0, 1].set_title("Plot of Double")

axis[1, 0].plot(x, y3, label='y = x^2', color='green')
axis[1, 0].set_title("Plot of Square")

axis[1, 1].plot(x, y4, label='y = x^3', color='red')
axis[1, 1].set_title("Plot of Cube")

fig.suptitle("Various Mathematical Functions", fontsize=16)
fig.tight_layout()
plt.show()
```

```python
# Temperature over cities
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temps = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]

fig, ax = plt.subplots(2, 2)

n = 0
for i in range(2):
    for j in range(2):
        ax[i, j].plot(days, temps[n])
        ax[i, j].set_title(cities[n])
        ax[i, j].grid(True)
        n += 1

fig.suptitle("Temperature Variation Over Cities", fontsize=16)
fig.supylabel("Temperature (°C)", fontsize=12)
fig.supxlabel("Days of the Week", fontsize=12)
fig.tight_layout()
plt.show()
```

```python
# Temperature over cities
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temps = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]

fig, ax = plt.subplots(2, 2)

ax = ax.flatten()

for i, ax in enumerate(ax):
    ax.plot(days, temps[i])
    ax.set_title(cities[i])
    ax.grid(True)

fig.suptitle("Temperature Variation Over Cities", fontsize=16)
fig.supylabel("Temperature (°C)", fontsize=12)
fig.supxlabel("Days of the Week", fontsize=12)
fig.tight_layout()
plt.show()
```

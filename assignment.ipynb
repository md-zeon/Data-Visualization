```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
scores = np.random.normal(loc=70, scale=10, size=200)
print(scores)
```

```python
sns.histplot(data=scores, bins=20)
plt.axvline(x=np.mean(scores), color='red', linestyle='dashed', linewidth=1, label=f'Mean: {np.mean(scores):.2f}')
plt.title("Distribution of Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex')
sns.set_style("darkgrid")
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='smoker', size='size',)
sns.set_style("white")
```

```python
pivot_table = tips.pivot_table(values='tip', index='day', columns='time', aggfunc='mean')
print(pivot_table)
plt.figure(figsize=(8, 6))
sns.heatmap(pivot_table, annot=True, fmt=".2f", cmap="YlGnBu", cbar_kws={'label': 'Average Tip'})
plt.title("Average Tip by Day and Time")
plt.xlabel("Time")
plt.ylabel("Day")
plt.tight_layout()
plt.show()
```
